## 2. Main Lab Configuration

These are the stable Spectrum hardware settings for the present setup.

Most experiment-specific RF parameters remain in the corresponding test sections below.


```python
DEVICE = "/dev/spcm0"
EXPECTED_MODEL = "M4i.6622-x8"
EXPECTED_SERIAL = 23161

CHANNEL = spcm.CHANNEL0
LOAD = 50 * units.ohm
CHANNEL_RANGE = 0.5 * units.V

print("Device:", DEVICE)
print("Expected model:", EXPECTED_MODEL)
print("Expected serial:", EXPECTED_SERIAL)
```

### Reusable Spectrum Setup Helpers

The repeated card / channel / DDS initialization is kept in a few small helper functions.

The actual DDS commands for each experiment remain visible in the experiment cells.


```python
def open_dds():
    card = spcm.Card(DEVICE).open()

    try:
        if card.product_name() != EXPECTED_MODEL or card.sn() != EXPECTED_SERIAL:
            raise RuntimeError(
                f"Unexpected card: {card.product_name()}, serial {card.sn()}"
            )

        card.card_mode(spcm.SPC_REP_STD_DDS)

        channels = spcm.Channels(
            card,
            card_enable=CHANNEL,
        )

        channels[0].enable(True)
        channels[0].output_load(LOAD)
        channels[0].amp(CHANNEL_RANGE)

        card.write_setup()

        dds = spcm.DDS(
            card,
            channels=channels,
        )
        dds.reset()

        return card, dds

    except Exception:
        card.close()
        raise


def start_output(card):
    card.start(
        spcm.M2CMD_CARD_ENABLETRIGGER,
        spcm.M2CMD_CARD_FORCETRIGGER,
    )


def stop_output(card):
    try:
        card.stop()
    finally:
        card.close()
```

## 3. Card Connection Check

Opens the card, reads its identity, verifies the model and serial number, then closes it.

No DDS configuration and no RF output are used here.

## 8. Multi-Tone Scope Test — Settings

Only use this after the single-tone test works.

Start with two tones, then increase gradually.

Each frequency uses one DDS core.


```python
MULTI_FREQUENCIES = [
    100 * units.MHz,
    110 * units.MHz,
]

MULTI_TOTAL_AMPLITUDE = 10 * units.percent
MULTI_PHASE = 0 * units.degrees

print("Number of tones:", len(MULTI_FREQUENCIES))
print("Frequencies:", MULTI_FREQUENCIES)
print("Total DDS amplitude budget:", MULTI_TOTAL_AMPLITUDE)
```

## 9. Multi-Tone RF Output

This cell assigns one DDS core to each programmed frequency.

Use the RIGOL built-in FFT to verify that all programmed tones are present. The FFT amplitude display can remain in **Vrms**.


```python
card, dds = open_dds()

if len(MULTI_FREQUENCIES) > len(dds):
    raise ValueError(
        f"Requested {len(MULTI_FREQUENCIES)} tones, "
        f"but only {len(dds)} DDS cores are available."
    )

amp_each = MULTI_TOTAL_AMPLITUDE / len(MULTI_FREQUENCIES)

for i, frequency in enumerate(MULTI_FREQUENCIES):
    dds[i].freq(frequency)
    dds[i].amp(amp_each)
    dds[i].phase(MULTI_PHASE)

    print(
        f"Core {i}: "
        f"{dds[i].get_freq(return_unit=units.MHz)}, "
        f"{dds[i].get_amp(return_unit=units.percent)}"
    )

dds.exec_at_trg()
dds.write_to_card()
start_output(card)

print("Multi-tone RF output ON")
print("Check the programmed peaks directly on the RIGOL FFT.")
```

### 10.1 Test A — Live Frequency Jump

The Spectrum card remains running while core 0 is moved through several discrete frequencies.

Each frequency is held long enough to observe manually on the RIGOL.

The update path is:

`set new frequency → exec_now() → write_to_card() → hold`

No waveform acquisition or Python FFT is used.


```python
FREQ_MOVE_POINTS_MHZ = [95, 99, 103, 107, 111]

FREQ_MOVE_AMPLITUDE = 100 * units.mV
FREQ_MOVE_PHASE = 0 * units.degrees

# Human-observation hold time on the RIGOL
FREQ_MOVE_HOLD_TIME_S = 3.0

print("Frequency setpoints [MHz]:", FREQ_MOVE_POINTS_MHZ)
print("Hold time per frequency:", FREQ_MOVE_HOLD_TIME_S, "s")
```

```python
card, dds = open_dds()

dds.trg_src(spcm.SPCM_DDS_TRG_SRC_NONE)

dds[0].amp(FREQ_MOVE_AMPLITUDE)
dds[0].phase(FREQ_MOVE_PHASE)
dds[0].freq(FREQ_MOVE_POINTS_MHZ[0] * units.MHz)

dds.exec_at_trg()
dds.write_to_card()

try:
    start_output(card)
    print("RF OUTPUT ON")

    for i, f_MHz in enumerate(FREQ_MOVE_POINTS_MHZ):
        if i:
            dds[0].freq(f_MHz * units.MHz)
            dds.exec_now()
            dds.write_to_card()

        print(f"Current frequency: {f_MHz:.3f} MHz")
        time.sleep(FREQ_MOVE_HOLD_TIME_S)

finally:
    stop_output(card)
    print("RF OUTPUT OFF")
```

### 10.2 Test B — DDS Timer-Triggered Frequency Steps

Several frequencies are queued before the card starts. The DDS internal timer then advances from one frequency to the next automatically.

Watch the frequency changes directly on the RIGOL FFT.


```python
TIMED_FREQUENCIES_MHZ = [95, 103, 111]
TIMED_STEP_PERIOD_S = 3.0
TIMED_AMPLITUDE = 100 * units.mV

card, dds = open_dds()

dds.trg_src(spcm.SPCM_DDS_TRG_SRC_TIMER)
dds.trg_timer(TIMED_STEP_PERIOD_S * units.s)
dds[0].amp(TIMED_AMPLITUDE)

for f_MHz in TIMED_FREQUENCIES_MHZ:
    dds[0].freq(f_MHz * units.MHz)
    dds.exec_at_trg()

dds.write_to_card()

print("Timed sequence:", " -> ".join(f"{f} MHz" for f in TIMED_FREQUENCIES_MHZ))
print("Step period:", TIMED_STEP_PERIOD_S, "s")

try:
    start_output(card)
    print("TIMED FREQUENCY MOVEMENT ON")
    time.sleep(TIMED_STEP_PERIOD_S * len(TIMED_FREQUENCIES_MHZ))
finally:
    stop_output(card)
    print("RF OUTPUT OFF")
```

### 10.3 Test C — Linear Frequency Ramp

This uses Spectrum DDS `frequency_slope()` to produce a continuous hardware ramp.

Watch the peak move continuously on the RIGOL FFT.

The sequence is:

1. output the start frequency;
2. enable the linear frequency slope;
3. ramp to the stop frequency;
4. stop the slope and hold the final frequency.


```python
RAMP_START_MHZ = 95
RAMP_STOP_MHZ = 111
RAMP_TIME_S = 5.0
RAMP_FINAL_HOLD_S = 2.0

RAMP_AMPLITUDE = 100 * units.mV
RAMP_STEP_DIVIDER = 1000

slope = (
    (RAMP_STOP_MHZ - RAMP_START_MHZ)
    * units.MHz
    / (RAMP_TIME_S * units.s)
)

card, dds = open_dds()

dds.trg_src(spcm.SPCM_DDS_TRG_SRC_TIMER)
dds.trg_timer(RAMP_TIME_S * units.s)
dds.freq_ramp_stepsize(RAMP_STEP_DIVIDER)

dds[0].amp(RAMP_AMPLITUDE)
dds[0].freq(RAMP_START_MHZ * units.MHz)
dds.exec_at_trg()

dds[0].frequency_slope(slope)
dds.exec_at_trg()

dds[0].frequency_slope(0 * units.MHz / units.s)
dds[0].freq(RAMP_STOP_MHZ * units.MHz)
dds.exec_at_trg()

dds.write_to_card()

print(f"Ramp: {RAMP_START_MHZ} -> {RAMP_STOP_MHZ} MHz")
print("Ramp time:", RAMP_TIME_S, "s")
print("Frequency slope:", slope)

try:
    start_output(card)
    print("LINEAR FREQUENCY RAMP ON")
    time.sleep(2 * RAMP_TIME_S + RAMP_FINAL_HOLD_S)
finally:
    stop_output(card)
    print("RF OUTPUT OFF")
```

### 10.4 Linear Frequency-Movement Animation

This animation visualizes the **programmed continuous linear trajectory** from Test C.

It does not read data from the RIGOL and does not control the hardware. It is only a software visualization of the programmed DDS frequency versus time.


```python
LINEAR_ANIMATION_FRAMES = 120
LINEAR_ANIMATION_INTERVAL_MS = 40
LINEAR_ANIMATION_REPEAT = True

linear_t = np.linspace(
    0,
    RAMP_TIME_S,
    LINEAR_ANIMATION_FRAMES,
)

linear_f = np.linspace(
    RAMP_START_MHZ,
    RAMP_STOP_MHZ,
    LINEAR_ANIMATION_FRAMES,
)

fig, (ax_freq, ax_time) = plt.subplots(
    2,
    1,
    figsize=(9, 6),
    gridspec_kw={"height_ratios": [1, 2]},
)

margin = max(
    1.0,
    0.1 * abs(RAMP_STOP_MHZ - RAMP_START_MHZ),
)

ax_freq.set_xlim(
    min(RAMP_START_MHZ, RAMP_STOP_MHZ) - margin,
    max(RAMP_START_MHZ, RAMP_STOP_MHZ) + margin,
)
ax_freq.set_ylim(0, 1)
ax_freq.set_yticks([])
ax_freq.set_xlabel("Instantaneous RF Frequency [MHz]")
ax_freq.set_title("Linear Frequency Movement")
ax_freq.grid(True, axis="x")

freq_line = ax_freq.axvline(
    RAMP_START_MHZ,
    linewidth=3,
)

freq_text = ax_freq.text(
    0.02,
    0.82,
    "",
    transform=ax_freq.transAxes,
)

ax_time.plot(
    linear_t,
    linear_f,
    linewidth=2,
    label="Programmed linear ramp",
)

current_point, = ax_time.plot(
    [linear_t[0]],
    [linear_f[0]],
    "o",
    markersize=8,
)

ax_time.set_xlim(0, RAMP_TIME_S)
ax_time.set_ylim(
    min(RAMP_START_MHZ, RAMP_STOP_MHZ) - margin,
    max(RAMP_START_MHZ, RAMP_STOP_MHZ) + margin,
)
ax_time.set_xlabel("Time [s]")
ax_time.set_ylabel("Frequency [MHz]")
ax_time.grid(True)
ax_time.legend()

def update_linear_frequency(frame):
    t_now = linear_t[frame]
    f_now = linear_f[frame]

    freq_line.set_xdata([f_now, f_now])
    freq_text.set_text(
        f"t = {t_now:.3f} s\n"
        f"f = {f_now:.3f} MHz"
    )

    current_point.set_data(
        [t_now],
        [f_now],
    )

    return freq_line, freq_text, current_point


linear_frequency_animation = FuncAnimation(
    fig,
    update_linear_frequency,
    frames=LINEAR_ANIMATION_FRAMES,
    interval=LINEAR_ANIMATION_INTERVAL_MS,
    repeat=LINEAR_ANIMATION_REPEAT,
)

plt.close(fig)
display(HTML(linear_frequency_animation.to_jshtml()))
```

### 10.5 Test D — Complex Frequency Movement with DDS Command Queue

This section follows the control pattern of Spectrum's official `06_dds_time_continuous_changes.py`.

A non-monotonic arbitrary frequency path is streamed through the DDS command queue.

The important mechanisms are:

1. `SPCM_DDS_TRG_SRC_TIMER` generates regular DDS trigger events.
2. Frequency commands are queued using `freq(...)` + `exec_at_trg()`.
3. `queue_cmd_max()` gives the queue capacity.
4. `queue_cmd_count()` is monitored while running.
5. New blocks are added before the queue becomes empty.
6. `status()` is checked for queue underrun.

Watch the arbitrary frequency path directly on the RIGOL FFT.

Use **Ctrl+C / Interrupt Kernel** to stop the streaming test. The `finally` block stops and closes the Spectrum card.


```python
COMPLEX_FREQUENCIES_MHZ = [
    95,
    99,
    107,
    103,
    111,
    101,
    109,
    97,
    103,
]

COMPLEX_STEP_PERIOD_S = 1.0
COMPLEX_AMPLITUDE = 100 * units.mV
COMPLEX_PHASE = 0 * units.degrees

print("Complex frequency path [MHz]:")
print(COMPLEX_FREQUENCIES_MHZ)
print("Step period:", COMPLEX_STEP_PERIOD_S, "s")

complex_t = (
    np.arange(len(COMPLEX_FREQUENCIES_MHZ))
    * COMPLEX_STEP_PERIOD_S
)

plt.figure(figsize=(9, 4))
plt.step(
    complex_t,
    COMPLEX_FREQUENCIES_MHZ,
    where="post",
)
plt.plot(
    complex_t,
    COMPLEX_FREQUENCIES_MHZ,
    "o",
)
plt.xlabel("Time [s]")
plt.ylabel("DDS Frequency [MHz]")
plt.title("One Cycle of the Complex Frequency Path")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
card, dds = open_dds()

dds.trg_src(spcm.SPCM_DDS_TRG_SRC_TIMER)
dds.trg_timer(COMPLEX_STEP_PERIOD_S * units.s)

dds[0].amp(COMPLEX_AMPLITUDE)
dds[0].phase(COMPLEX_PHASE)
dds[0].freq(COMPLEX_FREQUENCIES_MHZ[0] * units.MHz)

dds.exec_at_trg()
dds.write_to_card()

queue_max = dds.queue_cmd_max()
fill_number = max(1, int(queue_max / 4))
fill_check = queue_max - fill_number

print("DDS queue maximum:", queue_max)
print("Commands per refill block:", fill_number)
print("Pre-filling DDS command queue...")

column = 1

for _ in range(fill_number):
    f_MHz = COMPLEX_FREQUENCIES_MHZ[
        column % len(COMPLEX_FREQUENCIES_MHZ)
    ]

    dds[0].freq(f_MHz * units.MHz)
    dds.exec_at_trg()
    column += 1

dds.write_to_card()

try:
    start_output(card)

    print("COMPLEX FREQUENCY MOVEMENT ON")
    print("Watch the RIGOL FFT.")
    print("Press Ctrl+C / Interrupt to stop.")

    while True:

        while dds.queue_cmd_count() >= fill_check:
            time.sleep(0.001)

        for _ in range(fill_number):
            f_MHz = COMPLEX_FREQUENCIES_MHZ[
                column % len(COMPLEX_FREQUENCIES_MHZ)
            ]

            dds[0].freq(f_MHz * units.MHz)
            dds.exec_at_trg()
            column += 1

        dds.write_to_card()

        if dds.status() & spcm.SPCM_DDS_STAT_QUEUE_UNDERRUN:
            print("ERROR: DDS command queue underrun")
            break

except KeyboardInterrupt:
    print("Complex frequency movement interrupted by user.")

finally:
    stop_output(card)
    print("RF OUTPUT OFF")
```

## 14. Quick Troubleshooting

### `import spcm` fails
Run the `%pip install spcm` cell at the top and confirm the package is installed in the same Jupyter environment.

### Card cannot open
Check:
- Spectrum driver
- `/dev/spcm0`
- card permissions
- Spectrum Control Center

### RIGOL shows no signal
Check:
- physical connection from Spectrum CH0;
- 50 Ω RF measurement configuration;
- scope channel enabled;
- appropriate time scale / vertical scale;
- RF output cell actually reached `RF output ON`.

### RIGOL FFT looks poor
Check:
- FFT center / span;
- record length / acquisition settings;
- vertical scale;
- input termination;
- FFT amplitude display.

For the present setup, **Vrms has produced a more useful FFT display than dBV**, so use Vrms unless there is a specific reason to switch units.

### Before connecting the AOD
Confirm:
- AOD operating frequency information;
- RF amplifier gain;
- amplifier output power;
- cable / coupler loss;
- final RF power delivered to the AOD remains below its allowed limit.
