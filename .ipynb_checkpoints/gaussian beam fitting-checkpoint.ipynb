```python
# load packages
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.optimize import curve_fit
from functools import partial
```

```python
# readout data from csv file
root_dir = f'X:\\migratedData\\Rydberg_QIS\\data\\2026_07_30 780 exc beam analysis\\'
file_name = str('test.csv') # replace with file name
file_path = root_dir + file_name

df = pd.read_csv(file_path, skiprows = 0, usecols = [1,3])
# df = pd.read_csv(file_path, skiprows = ?, usecols = [], encoding = 'latin1') # try this if 1st line doesn't work
data = df.to_numpy()

# replace ?'s with correct column from csv file
z_vals, w_vals = data[:,0], data[:,1]
```

```python
# define beam waist as a function of distance
# express everything in meters
def waist(z, w0, z0, lmbda):
    arg = w0**2 + ((z - z0)*lmbda/(np.pi*w0))**2
    return np.sqrt(arg)
```

```python
# fix parameter 'lmbda' explicitly
fixed_model = partial(waist, lmbda = 780e-09)

# when using partial with keyword arguments, you MUST provide an initial guess (p0)
# this lets curve_fit know exactly how many parameters are left to optimize
initial_guess = [0.1,1.0] # replace with guesses for 'w0' and 'z0'

popt, _ = curve_fit(fixed_model, z_vals, w_vals, p0 = initial_guess, maxfev = 10000000, method = 'lm')
print("Optimized [w0, z0]:", popt)
```

```python
z_test = np.linspace(-1,1,100)

plt.figure(1)
plt.scatter(z_vals*1e+02, w_vals*1e+03)
plt.plot(z_test*1e+02, waist(z_test,*popt,lmbda = 780e-09)*1e+03)
plt.xlabel('Distance from mirror [cm]')
plt.ylabel('Beam spot size $w_{0}$ [mm]')
plt.text(-100, 1.2, f'w0: {popt[0]*1e+03} mm \n z0: {popt[1]} m')
plt.show()
```
