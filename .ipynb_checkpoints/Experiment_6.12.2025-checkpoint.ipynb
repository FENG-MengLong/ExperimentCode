# Experiments

## Redberg Spectroscopy

### Parameters

```python
# set the pulse sequence
#                   channel width  delay
pulse_arrangement = [["A", 9000e-6, 0], # MOT
                     ["B", 4e-6, 9001e-6], # 480 EXC
                     ["C", 4e-6, 9001e-6], # 780 EXC
                     ["G", 50e-6, 9006e-6], # SR400 trigger, the width is not the gate width, it is just a trigger, it can also be scanned. To realize, scan it in the loop
                    ]

# test_pulse = [["G",5e-6,0]]
# photon counter gate width
gate_width = 50e-6

# period of a sequence
T = 10e-3

# number of the sequence cycle
cycle_number = 10

# frequency per motor step
freq_step = 78.5 #KHz

# motor step range
motor_step_range = np.arange(0,101)
relative_freq_range = freq_step * motor_step_range

# electrode setting
board_num = 0
channels = 10
bipolar = True
voltage_range = np.linspace(-5,5,101)
```

```python
# sr400.flush_stale_response()
```

```python
#set the photon counter
# sr400.counter_set(count_mode = "INDEPENDENT", count_preset = 1, count_period_num=cycle_number, dwell=T, sourceA="INPUT1",sourceB="INPUT2", gate_A_mode="FIXED", gate_A_delay=0,
#                 gate_A_width=gate_width,gate_B_mode="CW")
sr400.counter_set(count_mode = "INDEPENDENT", count_preset = 9.6e-3, count_period_num=cycle_number, dwell=0, sourceA="INPUT1",sourceB="INPUT2", gate_A_mode="FIXED", gate_A_delay=0,
                gate_A_width=gate_width,gate_B_mode="CW")
```

```python
bnc575.disable_all()
```

```python
bnc575.disarm_all()
# bnc575.clock_set(period=T, mode="BURST", burstctr=cycle_number)
bnc575.clock_set(period=T,mode="CONTINUOUS")

# bnc575.channel_set(channel="H",width=pulse_arrangement[0][1],delay=pulse_arrangement[0][2], MUX="00001111")

for idn, pulse in enumerate(pulse_arrangement):
    bnc575.channel_set(channel=pulse[0],width=pulse[1],delay=pulse[2])
```

```python
v=0
# mcbox.configure(board_num, channels)
mcbox.set_voltage_1chan(board_num, channels, v, bipolar = False)
# mcbox.set_voltage_1chan(board_num, channels+1, v, bipolar = True)
```

```python
# result list
photon_num_list = []
bnc575.start_pulses()
for step in motor_step_range:
    counts_one_freq = []
    for voltage in voltage_range:
        mcbox.set_voltage_1chan(board_num, channels, voltage, bipolar= True)
        time.sleep(0.01)
        sr400.count_reset()
        time.sleep(0.01)
        counts_one_freq.append(sr400.read_entire_counts(counter="A",length = cycle_number))
        weeder.step(header="A",direction="+")
        time.sleep(0.05)
    photon_num_list.append(counts_one_freq)

bnc575.disarm_all()
```
