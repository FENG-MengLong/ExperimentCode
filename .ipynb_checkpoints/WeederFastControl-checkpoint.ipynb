# speed test (move 1000 steps)

## move

```python
s = time.time()
weeder.pyvisa.write("A" + "M" + str(4000) + "\r")
e = time.time()
print(e-s)
```

## step iteration

```python
s = time.time()
for i in range(1000):
    weeder.step("A","+")
e = time.time()
print(e-s)
```
