```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the data (your file has two columns, whitespace-separated)
file_path = 'C:/Users/raithel-lab/Desktop/New folder/0/20251128_17563355_peaks_C1.dat'
df = pd.read_csv(file_path, delim_whitespace=True, header=None)
df.columns = ['x', 'y']   # rename columns for clarity

# Plot
plt.figure(figsize=(8, 5))
plt.plot(df['x'], df['y'])
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Data Plot from peaks_C1.dat')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import glob

folder = "C:/Users/raithel-lab/Desktop/New folder/2000/"       # directory containing your .dat files
files = glob.glob(folder + "*.dat")

plt.figure(figsize=(8,5))

for f in files:
    df = pd.read_csv(f, delim_whitespace=True, header=None)
    df.columns = ['x', 'y']
    plt.plot(df['x'], df['y'], label=f.split('/')[-1])

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Plot of all .dat files")
plt.legend()
plt.grid(True)
plt.show()
```
