```python
# import packages
import numpy as np
import pyvisa
import asyncio
import matplotlib.pyplot as plt
```

### Searching for and identifying devices

```python
# looking for connected, interfaceable (?) devices
rm = pyvisa.ResourceManager()
print(rm.list_resources())
```

```python
# the 9350A Lecroy oscilloscope supports the *IDN? query command and will return with identifying information
lecroy = rm.open_resource('GPIB0::4::INSTR')
print(lecroy.query('*IDN?'))
```

```python
# conversely, the SR400 photon counter does not support the *IDN? query command and will return an error
sr400 = rm.open_resource('GPIB0::23::INSTR')
print(sr400.query('*IDN?'))
```

### Basic controls on Lecroy oscilloscope

```python
# check current timebase
print(lecroy.query('TIME_DIV?'))

# set, read out new timebase
lecroy.write('TIME_DIV 5 US')
print(lecroy.query('TIME_DIV?'))

#'''
#some commands don't appear to be case-sensitive
#print(lecroy.query('time_div?'))
#lecroy.write('time_div 20 us')
#'''

# changing voltage offset on a channel; again is not case-sensitive (for multiple commands, separate with a semicolon)
lecroy.write("c2:ofst 1 v; c1:ofst 0 mv")

# checking the vertical sensitivity and offset on a scope channel
print(lecroy.query("C2:VDIV?; C2:OFST?"))

# check attenuation factor
print(lecroy.query("c1:attenuation?"))
```
