```python
trigger_src = "/Dev1/PFI0"

task0 = nidaqmx.Task()
task0.co_channels.add_co_pulse_chan_time("Dev1/ctr0",idle_state=Level.LOW,initial_delay = 0, high_time=0.05e-6,low_time=0.05e-6)
task0.timing.cfg_implicit_timing(sample_mode = AcquisitionType.FINITE,samps_per_chan=1)
task0.triggers.start_trigger.cfg_dig_edge_start_trig(trigger_source=trigger_src,trigger_edge=Edge.RISING)

task1 = nidaqmx.Task()
task1.co_channels.add_co_pulse_chan_time("Dev1/ctr1",idle_state=Level.LOW, initial_delay = 0.025e-6, high_time=0.05e-6,low_time=0.05e-6)
task1.timing.cfg_implicit_timing(sample_mode = AcquisitionType.FINITE,samps_per_chan=1)
task1.triggers.start_trigger.cfg_dig_edge_start_trig(trigger_source=trigger_src,trigger_edge=Edge.RISING)

task0.start()
task1.start()

input("Generating pulse train. Press Enter to stop.\n")
task0.close()
task1.close()

```

```python
trigger_src = "/Dev1/PFI0"

task0 = nidaqmx.Task()
task0.co_channels.add_co_pulse_chan_time("Dev1/ctr0",idle_state=Level.LOW,initial_delay = 0, high_time=5e-3,low_time=5e-3)
task0.timing.cfg_implicit_timing(sample_mode = AcquisitionType.FINITE,samps_per_chan=1)
task0.triggers.start_trigger.cfg_dig_edge_start_trig(trigger_source=trigger_src,trigger_edge=Edge.RISING)

task00 = nidaqmx.Task()
task00.ci_channels.add_ci_pulse_width_chan(counter="Dev1/ctr2",min_val=1e-6,max_val=1.0,starting_edge=Edge.RISING)
task00.timing.cfg_implicit_timing(sample_mode=AcquisitionType.FINITE,samps_per_chan=2)
task00.start()

task1 = nidaqmx.Task()
task1.co_channels.add_co_pulse_chan_time("Dev1/ctr1",idle_state=Level.LOW, initial_delay = 0, high_time=5e-3,low_time=5e-3)
task1.timing.cfg_implicit_timing(sample_mode = AcquisitionType.FINITE,samps_per_chan=1)
task1.triggers.start_trigger.cfg_dig_edge_start_trig(trigger_source=trigger_src,trigger_edge=Edge.RISING)

task11 = nidaqmx.Task()
task11.ci_channels.add_ci_pulse_width_chan(counter="Dev1/ctr3",min_val=1e-6,max_val=1.0,starting_edge=Edge.RISING)
task11.timing.cfg_implicit_timing(sample_mode=AcquisitionType.FINITE,samps_per_chan=2)
task11.start()

task0.start()
task1.start()

t00 = task00.read()
t11 = task11.read()
input("Generating pulse train. Press Enter to stop.\n")
print(f"pulse width is: {t11}")
task0.close()
task1.close()
task00.close()
task11.close()
```

# Pulse Train test

```python
trigger_src = "/Dev1/PFI0"

task0 = nidaqmx.Task()
task0.co_channels.add_co_pulse_chan_freq("Dev1/ctr0", idle_state=Level.LOW, initial_delay=0.1e-6, freq=1e6, duty_cycle=0.5)
task0.timing.cfg_implicit_timing(sample_mode=AcquisitionType.CONTINUOUS)
task0.triggers.start_trigger.cfg_dig_edge_start_trig(trigger_source=trigger_src, trigger_edge=Edge.RISING)

task1 = nidaqmx.Task()
task1.co_channels.add_co_pulse_chan_freq("Dev1/ctr1", idle_state=Level.LOW, initial_delay=0, freq=1e6, duty_cycle=0.5)
task1.timing.cfg_implicit_timing(sample_mode=AcquisitionType.CONTINUOUS)
task1.triggers.start_trigger.cfg_dig_edge_start_trig(trigger_source=trigger_src, trigger_edge=Edge.RISING)

task0.start()
task1.start()

input("Generating pulse train. Press Enter to stop.\n")

task0.close()
task1.close()
```
