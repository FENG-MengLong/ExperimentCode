```python
# source (command list starts on pg. 62): https://cdn.teledynelecroy.com/files/manuals/9300-rcm_reva.pdf
class lecroy:
    def __init__(self, visa_name: str, timeout = 5000):
        rm = pyvisa.ResourceManager()
        self.pyvisa = rm.open_resource(visa_name)
        self.pyvisa.timeout = 5000 # couldn't we just set this to timeout parameter?

    # the functions below are used to set trigger paramters
    def trig_coupling(self, trig_source: str, trig_coupling: str, query = False):
        # command or query, pg. 216
        if query:
            print(self.pyvisa.query(trig_source + ":TRCP?"))
        else:
            self.pyvisa.write(trig_source + ":TRCP " + trig_coupling)
            
    def trig_delay(self, trig_delay: str, query = False):
        # command or query, pg. 217
        if query:
            print(self.pyvisa.query("TRDL?"))
        else:
            self.pyvisa.write("TRDL " + delay)
            
    def trig_level(self, trig_source: str, trig_level: str, query = False):
        # command or query, pg. 218
        if query:
            print(self.pyvisa.query(trig_source + ":TRLV?"))
        else:
            self.pyvisa.write(trig_source + ":TRLV " + trig_level)
            
    def trig_mode(self, trig_mode: str, query = False):
        # command or query, pg. 219
        if query:
            print(self.pyvisa.query("TRMD?"))
        else:
            self.pyvisa.write("TRMD " + trig_mode)
            
    """def trig_select(query = False):  # WORK ON THIS
        # command or query, pg. 222
        if query:
            print(self.pyvisa.query("TRSE?"))
        else:
            self.pyvisa.write()"""
            
    def trig_slope(self, trig_source: str, trig_slope: str, query = False):
        # command or query, pg. 225
        if query:
            print(self.pyvisa.query(trig_source + ":TRSL?"))
        else:
            self.pyvisa.write(trig_source + ":TRSL " + trig_slope)
            
    def trig_window(self, value: str, query = False):
        # command or query, pg. 226
        if query:
            print(self.pyvisa.query("TRWI?"))
        else:
            self.pyvisa.write("TRWI " + value)
            
    # the functions below are used to set signal paramters

    def arm_acquisition(self):
        # command, pg. 73
        self.pyvisa.write("ARM")
        
    """def auto_setup(self, channel: str): ### CORRECT THIS
        # command, pg. 76
        # channel = "C1", "C2", etc.
        self.pyvisa.write(channel + ":ASET")"""
        
    def attenuation(self, channel: str, attn_factor, query = False):
        # command or query, pg. 74
        if query:
            print(self.pyvisa.query(channel + ":ATTN?"))
        else:
            self.pyvisa.write(channel + ":ATTN " + str(attn_factor))
            
    def bnc(self, channel: str, mode: str, query = False):
        # command or query, pg. 78
        if query:
            print(self.pyvisa.query(channel + ":BNC?"))
        else:
            self.pyvisa.write(channel + ":BNC " + mode)
            
    def bandwidth_limit(self, mode: str, query = False):
        # command or query, pg. 77
        if query:
            print(self.pyvisa.query("BWL?"))
        else:
            self.pyvisa.write("BWL " + mode)
        
    def coupling(self, channel: str, coupling: str, query = False):
        # command or query, pg. 101
        if query:
            print(self.pyvisa.query(channel + ":CPL?"))
        else:
            self.pyvisa.write(channel + ":CPL " + coupling)

    def hor_position(self, exp_trace: str, hor_position: str, segment: str, query = False):
        # command or query, pg. 144
        if query:
            print(self.pyvisa.query(exp_trace + ":HPOS?"))
        else:
            self.pyvisa.write(exp_trace + ":HPOS " + hor_position + " DIV," + segment)
            
    def offset(self, channel: str, offset: str, query = False):
        # command or query, pg. 160
        if query:
            print(self.pyvisa.query(channel + ":OFST?"))
        else:
            self.pyvisa.write(channel + ":OFST " + offset)
        
    def sample_clock(self, state: str, record_length: str, coupling: str, query = False):
        # command or query, pg. 197
        if query:
            print(self.pyvisa.query("SCLK?"))
        else:
            self.pyvisa.write("SCLK" + state + "," + record_length + "," + coupling)
            
    def stop(self):
        # command, pg. 206
        self.pyvisa.write("STOP")
        
    def time_div(self, time_div: str, query = False):
        # command or query, pg. 212
        if query:
            print(self.pyvisa.query("TDIV?"))
        else:
            self.pyvisa.write("TDIV " + time_div)

    def trace(self, trace: str, mode: str, query = False):
        # command or query, pg. 213
        if query:
            print(self.pyvisa.query(trace + ":TRA?"))
        else:
            self.pyvisa.write(trace + ":TRA " + mode)
        
    def vert_position(self, trace: str, display_offset: str, query = False):
        # command or query, pg. 230
        if query:
            print(self.pyvisa.query(trace + ":VPOS?"))
        else:
            self.pyvisa.write(trace + ":VPOS " + display_offset)
        
    def volt_div(self, channel: str, volt_div: str, query = False):
        # command or query, pg. 231
        if query:
            print(self.pyvisa.query(channel + ":VDIV?"))
        else:
            self.pyvisa.write(channel + ":VDIV " + volt_div)
        
# cls, wait, trg commands
# capturing and saving scope traces
# how to perform math on traces
# pg. 116

    # misc. functions:
    def message(self, message: str, query = False):
        # command or query, pg. 158
        # message can be up to 49 characters
        if query:
            print(self.pyvisa.query("MSG?"))
        else:
            self.pyvisa.write("MSG " + message)
    def wait(self):
        # command, pg. 233
        self.pyvisa.write("WAIT")
    def trg(self):
        # command, pg. 215
        self.pyvisa.write("*TRG")
    def cls(self):
        # command, pg. 86
        self.pyvisa.write("*CLS")

    def rst(self):
        # command, pg. 196
        self.pyvisa.write("*RST")

    """
    def define(self, trace: str, equation: str, max_points: int, sweeps: int, query = False):
        # command or query, pg. 111
        if query:
            print(self.pyvisa.query(trace + ":DEF?"))
        else:
            #self.pyvisa.write(trace + ":DEF EQN," + f"'{equation}'," + "MAXPTS," + f"{max_points}," + "SWEEPS," + f"{sweeps}")
            self.pyvisa.write(f'{trace}:DEF EQN,"{equation}",MAXPTS,{max_points},SWEEPS,{sweeps}')
    """

    def define(self, trace: str, equation: str, query = False, more = False, **options):
        try:
            if query:
                print(self.pyvisa.query(trace + ":DEF?"))
            else:
                command = f'{trace}:DEF EQN,"{equation}"'
                if more:
                    command += ','
                    num_vars = len(options)
                    for param, value in options.items():
                        command += f'{param},{value}'
                        num_vars -= 1
                        if num_vars > 0:
                            command += (',')
                        else:
                            continue   
                    self.pyvisa.write(command)
                else:
                    self.pyvisa.write(command)
                print(command)
        except Exception as e:
            print(f'An error occurred: {e}')
            
    # control file directories on scope
    # not sure if this scope supports this command
    def directory(self, device: str, action = None, directory = None, query = False):
        try:
            if query:
                if type(directory) == None:
                    print(self.pyvisa.query(f'DIR?DISK,{device}'))
                else:
                    print(self.pyvisa.query(f'DIR?DISK,{device},"{directory}"'))
            else:
                if type(action) == None:
                    return print(f'An error has occurred. No action parameter (CREATE, DELETE, SWITCH) given.')
                elif type(directory) == None:
                    return print(f'An error has occurred. No directory name given.')
                else:
                    self.pyvisa.write(f'DIRDISK,{device},ACTION,{action},"{directory}"')
        except Exception as e:
            print(f'An error occurred: {e}')

    def opt(self):
        print(self.pyvisa.query(f'*OPT?'))
```
