```python
# function for performing the pairwise voltage sweep
# use multithreading to set voltages simultaneously
def set_voltage_pairs(axis: str, voltage: float, bipolar = True):
    default_voltage = 0.0
    try:
        # used to control the x-electrodes, set all others to zero
        if axis == "X":
            X_channels = np.array([10,11],int)  # first channel is -x, second channel is +x
            fixed_channels = np.array([0,1,6,7],int)  # the rest of the electrodes will be set to zero
            task1 = threading.Thread(target = mcbox.set_voltage_1chan, 
                                     args = (board_num = 0, channel = X_channels[0], voltage = voltage, bipolar = True))
            task2 = threading.Thread(target = mcbox.set_voltage_1chan, 
                                     args = (board_num = 0, channel = X_channels[1], voltage = -voltage, bipolar = True))
            task3 = threading.Thread(target = mcbox.set_voltage_1chan, 
                                     args = (board_num = 0, channel = fixed_channels[0], voltage = default_voltage, bipolar = True))
            task4 = threading.Thread(target = mcbox.set_voltage_1chan, 
                                     args = (board_num = 0, channel = fixed_channels[1], voltage = default_voltage, bipolar = True))
            task5 = threading.Thread(target = mcbox.set_voltage_1chan, 
                                     args = (board_num = 0, channel = fixed_channels[2], voltage = default_voltage, bipolar = True))
            task6 = threading.Thread(target = mcbox.set_voltage_1chan, 
                                     args = (board_num = 0, channel = fixed_channels[3], voltage = default_voltage, bipolar = True))
            task1.start()
            task2.start()
            task3.start()
            task4.start()
            task5.start()
            task6.start()

            task1.join()
            task2.join()
            task3.join()
            task4.join()
            task5.join()
            task6.join()
        # used to control the y-electrodes, set all others to zero
        elif axis == "Y":
            Y_channels = np.array([6,7],int) # first channel is -y, second channel is +y
            fixed_channels = np.array([0,1,10,11],int) # the rest of the electrodes will be set to zero
            task1 = threading.Thread(target = mcbox.set_voltage_1chan, 
                                     args = (board_num = 0, channel = Y_channels[0], voltage = voltage, bipolar = True))
            task2 = threading.Thread(target = mcbox.set_voltage_1chan, 
                                     args = (board_num = 0, channel = Y_channels[1], voltage = -voltage, bipolar = True))
            task3 = threading.Thread(target = mcbox.set_voltage_1chan, 
                                     args = (board_num = 0, channel = fixed_channels[0], voltage = default_voltage, bipolar = True))
            task4 = threading.Thread(target = mcbox.set_voltage_1chan, 
                                     args = (board_num = 0, channel = fixed_channels[1], voltage = default_voltage, bipolar = True))
            task5 = threading.Thread(target = mcbox.set_voltage_1chan, 
                                     args = (board_num = 0, channel = fixed_channels[2], voltage = default_voltage, bipolar = True))
            task6 = threading.Thread(target = mcbox.set_voltage_1chan, 
                                     args = (board_num = 0, channel = fixed_channels[3], voltage = default_voltage, bipolar = True))
            task1.start()
            task2.start()
            task3.start()
            task4.start()
            task5.start()
            task6.start()

            task1.join()
            task2.join()
            task3.join()
            task4.join()
            task5.join()
            task6.join()
        # used to control the z-electrodes, set all others to zero
        elif axis == "Z":
            Z_channels = np.array([0,1],int) # first channel is -z, second channel is +z
            fixed_channels = np.array([6,7,10,11],int) # the rest of the electrodes will be set to zero
            task1 = threading.Thread(target = mcbox.set_voltage_1chan, 
                                     args = (board_num = 0, channel = Z_channels[0], voltage = default_voltage, bipolar = True))
            task2 = threading.Thread(target = mcbox.set_voltage_1chan, 
                                     args = (board_num = 0, channel = Z_channels[1], voltage = voltage, bipolar = True))
            task3 = threading.Thread(target = mcbox.set_voltage_1chan, 
                                     args = (board_num = 0, channel = fixed_channels[0], voltage = default_voltage, bipolar = True))
            task4 = threading.Thread(target = mcbox.set_voltage_1chan, 
                                     args = (board_num = 0, channel = fixed_channels[1], voltage = default_voltage, bipolar = True))
            task5 = threading.Thread(target = mcbox.set_voltage_1chan, 
                                     args = (board_num = 0, channel = fixed_channels[2], voltage = default_voltage, bipolar = True))
            task6 = threading.Thread(target = mcbox.set_voltage_1chan, 
                                     args = (board_num = 0, channel = fixed_channels[3], voltage = default_voltage, bipolar = True))
            task1.start()
            task2.start()
            task3.start()
            task4.start()
            task5.start()
            task6.start()

            task1.join()
            task2.join()
            task3.join()
            task4.join()
            task5.join()
            task6.join()
        else:
            raise RuntimeError("Invalid argument given for the axis. Valid options: 'X', 'Y', 'Z'.")
        return print('Multichannel assignment complete. Voltages successfully set.')
    except Exception as e:
        print(f'The following error occurred: {e}')
```

```python
# set the stepper motor scan range centered around a Rydberg resonance (e.g., 50D 3/2)
delta = 1000
center = 3400

# initialize the stepper motor position
start_position = center - delta
x_list = np.arange(2*delta)

# define test voltages and axis sweeps
axes = ["Z", "X", "Y"]
voltages = np.linspace(-1.0,1.0,11)  # go from -1 V to + 1 V in 0.2-volt steps

# create array to store mean counts versus electrode pair, compensation voltage, and stepper motor position/blue laser frequency
results = np.zeros((len(axes),len(voltages),len(x_list)))

# loop through the electrode pairs, stepping the voltage and then scanning the blue laser to record mean counts
i = 0
for axis in axes:
    j = 0
    for voltage in voltages:
        k = 0
        set_voltage_pairs(axis = axis, voltage = voltage, bipolar = True) # set voltages
        weeder.move(header = "A", position = start_position) # reset stepper motor
        # start pulse sequence/photon counting, iterate through stepper motor range, record mean number of counts
        for idx in x_list:
            print("Now checking step: {}".format(idx))
            sr400.count_reset()
            time.sleep(0.1)
            result = np.mean(sr400.read_entire_counts(counter = "A", length = cycle_number))
            results[i,j,k] = result
            weeder.advance(header = "A", num_step = 1)
            time.sleep(0.1)
            k += 1
        j += 1
    i += 1
```
